```python
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F

x = torch.linspace(-10, 5, 100)
ys = torch.sigmoid(x)
ye = torch.exp(x)
yre = torch.relu(x)
ysp = F.softplus(x)

plt.plot(x, ys, label='sigmoid')
# plt.plot(x, ye, label='exp')
plt.plot(x, yre, label='relu')
plt.plot(x, ysp, label='softplus')
plt.legend()
plt.show()
```
