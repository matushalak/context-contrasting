# Transition summaries

This notebook is intentionally concise:
- mean summaries only
- rotated 45 degree sectors instead of the old sign quadrants
- one figure style reused for `act`, `post/familiar`, and `post/novel`

```python
from pathlib import Path

import context_contrasting.data_analysis.transitions_helpers as th

build_mean_summary = th.build_mean_summary
DEFAULT_PLOT_STYLE = th.DEFAULT_PLOT_STYLE
compute_response_limits = th.compute_response_limits
compute_shift_limits = th.compute_shift_limits
export_figure_panels = th.export_figure_panels
load_transition_table = th.load_transition_table
plot_mean_transition_summary = th.plot_mean_transition_summary
sector_fraction_table = th.sector_fraction_table

PLOT_STYLE = DEFAULT_PLOT_STYLE | {
    "pre_point_alpha": 1.0,
    "target_point_alpha": 1.0,
    "shift_point_alpha": 1.0,
    "pre_vector_alpha": 0.5,
    "target_vector_alpha": 0.5,
    "individual_vector_width": 0.005,
    "mean_arrow_width": 3.1,
    "mean_arrow_mutation_scale": 16.5,
}

DATA_DIR = Path("context_contrasting/data_analysis")
if not DATA_DIR.exists():
    DATA_DIR = Path(".")

PANEL_EXPORT_DIR = DATA_DIR / "transition_panel_exports_gt_0_3"

act_table = load_transition_table(DATA_DIR / "transitions_act.csv")
post_table = load_transition_table(DATA_DIR / "transitions_post.csv")

# Threshold 0.3 works well
threshold = 0.3

act_mean = build_mean_summary(act_table, image_group="all", pre_stage="Pre", target_stage="Task", threshold=threshold)
post_familiar_mean = build_mean_summary(post_table, image_group="familiar", pre_stage="Pre", target_stage="Post", threshold=threshold)
post_novel_mean = build_mean_summary(post_table, image_group="novel", pre_stage="Pre", target_stage="Post", threshold=threshold)

post_response_lims = compute_response_limits(post_familiar_mean, post_novel_mean)
post_shift_lims = compute_shift_limits(post_familiar_mean, post_novel_mean)
```

## Act mean


```python
fig_act = plot_mean_transition_summary(
    act_mean,
    title="Transition mean summary - act",
    start_label="Pre",
    end_label="Task",
    style=PLOT_STYLE,
)
act_panel_paths = export_figure_panels(fig_act, PANEL_EXPORT_DIR, "act_mean")
act_sector_legend_paths = th.save_rotated_sector_unit_legend(
    act_mean,
    PANEL_EXPORT_DIR / "act_mean_sector_legend.png",
    label_fontsize=27
)
act_panel_paths + act_sector_legend_paths
```

```python
# post_familiar_mean = post_familiar_mean.loc[post_familiar_mean['dNorm'] > threshold]

fig_post_familiar = plot_mean_transition_summary(
    post_familiar_mean,
    title="Transition mean summary - post familiar",
    start_label="Pre",
    end_label="Post",
    response_lims=post_response_lims,
    shift_lims=post_shift_lims,
    style=PLOT_STYLE,
)
post_familiar_panel_paths = export_figure_panels(fig_post_familiar, PANEL_EXPORT_DIR, "post_familiar_mean")
post_familiar_sector_legend_paths = th.save_rotated_sector_unit_legend(
    post_familiar_mean,
    PANEL_EXPORT_DIR / "post_familiar_mean_sector_legend.png",
    label_fontsize=27
)
post_familiar_panel_paths + post_familiar_sector_legend_paths
```

```python
import numpy as np


def build_novel_no_familiar_o_summary(novel_summary, familiar_summary, *, target_label, threshold):
    summary = (
        novel_summary[["neuron_idx", "NO_Pre", "NO_Target"]]
        .merge(
            familiar_summary[["neuron_idx", "O_Pre", "O_Target"]],
            on="neuron_idx",
            how="inner",
        )
        .dropna(subset=["NO_Pre", "O_Pre", "NO_Target", "O_Target"])
        .copy()
    )
    summary["dNO"] = summary["NO_Target"] - summary["NO_Pre"]
    summary["dO"] = summary["O_Target"] - summary["O_Pre"]
    summary["dNorm"] = np.hypot(summary["dNO"].to_numpy(dtype=float), summary["dO"].to_numpy(dtype=float))
    summary["log_dNorm"] = np.log(summary["dNorm"] + th.LOG_NORM_EPS)
    summary = th.add_direction_columns(summary)
    summary = th.assign_rotated_sectors(summary, threshold=threshold)
    summary.attrs["pre_stage"] = "Naive"
    summary.attrs["target_stage"] = target_label
    summary.attrs["image_group"] = "novel NO vs familiar O"
    summary.attrs["transition"] = f"novel_no_familiar_o_{target_label.lower()}"
    return summary


def label_novel_no_familiar_o_summary(fig, *, target_label):
    axes = fig.axes
    target_lower = target_label.lower()

    for ax in (axes[0], axes[3], axes[6]):
        ax.set_xlabel("dNO (novel post - novel naive)")
        ax.set_ylabel(f"dO (familiar {target_lower} - familiar naive)")
    for ax in (axes[1], axes[4], axes[7]):
        ax.set_xlabel("NO response (novel, naive)")
        ax.set_ylabel("O response (familiar, naive)")
    for ax in (axes[2], axes[5], axes[8]):
        ax.set_xlabel("NO response (novel, post)")
        ax.set_ylabel(f"O response (familiar, {target_lower})")

    axes[0].set_title(f"{target_label} - naive by pre color")
    axes[1].set_title("Naive by pre color")
    axes[2].set_title(f"{target_label} by pre color")
    axes[3].set_title(f"{target_label} - naive by rotated sector")
    axes[4].set_title("Naive by rotated sector")
    axes[5].set_title(f"{target_label} by rotated sector")
    axes[6].set_title("|d| colored by coolwarm, alpha scaled by |d|")
    axes[7].set_title("Naive colored by |d| (coolwarm)")
    axes[8].set_title(f"{target_label} colored by |d| (coolwarm)")
    fig.tight_layout(rect=[0, 0.05, 1, 0.95])


familiar_task_o_vs_novel_no_mean = build_novel_no_familiar_o_summary(
    post_novel_mean,
    act_mean,
    target_label="Task",
    threshold=threshold,
)
familiar_expert_o_vs_novel_no_mean = build_novel_no_familiar_o_summary(
    post_novel_mean,
    post_familiar_mean,
    target_label="Expert",
    threshold=threshold,
)

mixed_response_lims = compute_response_limits(
    familiar_task_o_vs_novel_no_mean,
    familiar_expert_o_vs_novel_no_mean,
)
mixed_shift_lims = compute_shift_limits(
    familiar_task_o_vs_novel_no_mean,
    familiar_expert_o_vs_novel_no_mean,
)

fig_task_o_vs_novel_no = plot_mean_transition_summary(
    familiar_task_o_vs_novel_no_mean,
    title="Transition mean summary - familiar task O vs novel NO",
    start_label="Naive",
    end_label="Task",
    response_lims=mixed_response_lims,
    shift_lims=mixed_shift_lims,
    style=PLOT_STYLE,
)
label_novel_no_familiar_o_summary(fig_task_o_vs_novel_no, target_label="Task")

task_o_vs_novel_no_panel_paths = export_figure_panels(
    fig_task_o_vs_novel_no,
    PANEL_EXPORT_DIR,
    "familiar_task_occluded_vs_novel_nonoccluded_mean",
)
task_o_vs_novel_no_sector_legend_paths = th.save_rotated_sector_unit_legend(
    familiar_task_o_vs_novel_no_mean,
    PANEL_EXPORT_DIR / "familiar_task_occluded_vs_novel_nonoccluded_mean_sector_legend.png",
    label_fontsize=25
)
task_o_vs_novel_no_panel_paths + task_o_vs_novel_no_sector_legend_paths
```

```python
fig_expert_o_vs_novel_no = plot_mean_transition_summary(
    familiar_expert_o_vs_novel_no_mean,
    title="Transition mean summary - familiar expert O vs novel NO",
    start_label="Naive",
    end_label="Expert",
    response_lims=mixed_response_lims,
    shift_lims=mixed_shift_lims,
    style=PLOT_STYLE,
)
label_novel_no_familiar_o_summary(fig_expert_o_vs_novel_no, target_label="Expert")

expert_o_vs_novel_no_panel_paths = export_figure_panels(
    fig_expert_o_vs_novel_no,
    PANEL_EXPORT_DIR,
    "familiar_expert_occluded_vs_novel_nonoccluded_mean",
)
expert_o_vs_novel_no_sector_legend_paths = th.save_rotated_sector_unit_legend(
    familiar_expert_o_vs_novel_no_mean,
    PANEL_EXPORT_DIR / "familiar_expert_occluded_vs_novel_nonoccluded_mean_sector_legend.png",
    label_fontsize=25
)
expert_o_vs_novel_no_panel_paths + expert_o_vs_novel_no_sector_legend_paths
```
